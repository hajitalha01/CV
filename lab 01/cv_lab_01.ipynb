Task 1: Image Information
Load an image, display it, and print width, height, number of channels, data type, and number of pixels.


```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Load image
image = Image.open("/content/images.jfif")

# Display image
plt.imshow(image)
plt.axis('off')
plt.show()

# Convert image to NumPy array
img = np.array(image)

# Get image information
height, width = img.shape[:2]

if len(img.shape) == 2:
    channels = 1
else:
    channels = img.shape[2]

print("Width:", width)
print("Height:", height)
print("Number of Channels:", channels)
print("Data Type:", img.dtype)
print("Number of Pixels:", height * width)
```

Task 3: RGB Channels
Extract and separately display the red, green, and blue channels.


```python
# Extract RGB channels
red = img[:, :, 0]
green = img[:, :, 1]
blue = img[:, :, 2]

# Display channels separately
plt.figure(figsize=(12, 4))

# Red Channel
plt.subplot(1, 3, 1)
plt.imshow(red, cmap='gray')
plt.title("Red Channel")
plt.axis('off')

# Green Channel
plt.subplot(1, 3, 2)
plt.imshow(green, cmap='gray')
plt.title("Green Channel")
plt.axis('off')

# Blue Channel
plt.subplot(1, 3, 3)
plt.imshow(blue, cmap='gray')
plt.title("Blue Channel")
plt.axis('off')

plt.show()
```

Task 4: Image Manipulation
Crop the central region, resize to 256 × 256, convert to grayscale, and save the result.


```python
# 1. Get image dimensions
height, width = img.shape[:2]

# 2. Find central region
crop_size = min(height, width)

start_y = (height - crop_size) // 2
start_x = (width - crop_size) // 2

# 3. Crop central region
cropped = img[
    start_y:start_y + crop_size,
    start_x:start_x + crop_size
]

# 4. Resize to 256 × 256
resized = cv2.resize(cropped, (256, 256))

# 5. Convert to grayscale
if resized.shape[2] == 4:
    gray = cv2.cvtColor(resized, cv2.COLOR_RGBA2GRAY)
else:
    gray = cv2.cvtColor(resized, cv2.COLOR_RGB2GRAY)

# 6. Display result
plt.imshow(gray, cmap='gray')
plt.title("Final Grayscale Image")
plt.axis('off')
plt.show()

# 7. Save result
output = Image.fromarray(gray)
output.save("task4_result.jpg")

print("Image saved as task4_result.jpg")
```

Task 6: Quantization
Quantize a grayscale image to 256, 64, 16, 8, 4, and 2 intensity levels. Display and discuss the results.


```python
# Convert image to grayscale NumPy array
gray = np.array(gray_image)

# Function for quantization
def quantize_image(image, levels):
    step = 256 // levels
    quantized = (image // step) * step
    return quantized.astype(np.uint8)

# Create different quantized versions
q256 = quantize_image(gray, 256)
q64 = quantize_image(gray, 64)
q16 = quantize_image(gray, 16)
q8 = quantize_image(gray, 8)
q4 = quantize_image(gray, 4)
q2 = quantize_image(gray, 2)

# Display all results
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(q256, cmap='gray')
plt.title("256 Levels")
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(q64, cmap='gray')
plt.title("64 Levels")
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(q16, cmap='gray')
plt.title("16 Levels")
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(q8, cmap='gray')
plt.title("8 Levels")
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(q4, cmap='gray')
plt.title("4 Levels")
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(q2, cmap='gray')
plt.title("2 Levels")
plt.axis('off')

plt.tight_layout()
plt.show()
```
