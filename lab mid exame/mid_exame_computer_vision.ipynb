1. Specific object, region, or feature to detect, extract, or enhance
The main objective is to enhance the lung regions and detect their boundaries in the provided CT scan image. The two lungs are the primary regions of interest, while the heart and surrounding chest wall serve as nearby structures.

The project will focus on improving image contrast, reducing noise, and extracting the boundaries of the lungs using Gaussian filtering, Median filtering, and Canny edge detection.
2. Relevant image characteristics and challenges
- Noise: The CT scan contains small bright and dark variations that may interfere with accurate edge detection.
- Low contrast: Some lung structures and boundaries have similar intensity values, making them difficult to distinguish.
- Irregular boundaries: The lung boundaries are curved and uneven, requiring effective edge detection.
- Overlapping structures: The heart and other chest structures are close to the lungs, which may cause confusion during boundary extraction.
- Background variation: The image contains different intensity levels in the lungs, heart, chest wall, and surrounding background.

Median Filtering — Noise Removal


```python
# Load CT scan image
img = cv2.imread("/content/cmmm5208940-fig-0002a-m.png", cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError("ct_scan.png image not found!")

# Apply Median Filter
median_img = cv2.medianBlur(img, 3)

# Display original and filtered image
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original CT Scan")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(median_img, cmap="gray")
plt.title("Median Filter")
plt.axis("off")

plt.show()
```

Histogram Equalization — Contrast Enhancement

```python
# Apply Histogram Equalization
equalized_img = cv2.equalizeHist(img)

# Display original and enhanced image
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original CT Scan")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized_img, cmap="gray")
plt.title("Histogram Equalization")
plt.axis("off")

plt.show()
```

Canny Edge Detection — Edge Detection

```python
# Apply Canny Edge Detection
edges = cv2.Canny(
    median_img,
    threshold1=50,
    threshold2=150
)

# Display original and detected edges
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original CT Scan")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap="gray")
plt.title("Canny Edge Detection")
plt.axis("off")

plt.show()
```

Task no 04: Experiment Evaluation and comparison

```python
# Compare two Median Filter sizes
median_3 = cv2.medianBlur(img, 3)
median_5 = cv2.medianBlur(img, 5)

plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(median_3, cmap="gray")
plt.title("Median 3x3")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(median_5, cmap="gray")
plt.title("Median 5x5")
plt.axis("off")

plt.show()
```

```python
# Compare two Canny threshold settings
edges_1 = cv2.Canny(equalized_img, 30, 100)
edges_2 = cv2.Canny(equalized_img, 100, 200)

plt.figure(figsize=(12, 5))

plt.subplot(1, 3, 1)
plt.imshow(equalized_img, cmap="gray")
plt.title("Enhanced Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(edges_1, cmap="gray")
plt.title("Canny: 30, 100")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(edges_2, cmap="gray")
plt.title("Canny: 100, 200")
plt.axis("off")

plt.show()
```

The proposed techniques helped improve the CT scan image.

2. Limitations

3. Possible Improvements
